# Перевірка власноруч реалізованого алгоритму Apriori для пошуку асоціативних правил шляхом порівняння результатів із бібліотечним рішенням

```python
from mlxtend.frequent_patterns import apriori
from custom_apriory import apriori as custom_apriori
import pandas as pd

data = {'Milk': [1, 0, 1, 1, 0],
    'Bread': [1, 1, 0, 1, 1],
    'Butter': [0, 1, 1, 1, 0],
    'Cheese': [1, 0, 0, 1, 1]}

df = pd.DataFrame(data)
```

```python
frequent_itemsets_lib = apriori(df, min_support=0.5, use_colnames=True)

frequent_itemsets_custom = custom_apriori(df, min_support=0.5)

print("Lib Apriori:")
print(frequent_itemsets_lib)

print("Custom Apriori:")
print(frequent_itemsets_custom)
```

Власноручна реалізація алгоритма Apriori для пошуку асоціативних правил виконана успішно, результати повністю сходяться із бібліотечним рішенням.
